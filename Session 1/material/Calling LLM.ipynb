```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
```

## Loading Model

```python
model_name = "Qwen/Qwen2-1.5B-Instruct"

print("Loading model... (first time will download)")

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,   # use float16 if GPU
    device_map="auto"
)

print("Model loaded successfully!\n")
```

## LLM Function

```python
def ask_llm(prompt):
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=150,
        temperature=0.7
    )

    response = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return response
```

## Main Demo

```python
print("=== Offline Qwen Summarization Demo ===\n")

# Step 1: Input text
text = input("Enter text to summarize:\n")

# Step 2: Prompt engineering
prompt = f"""
You are an AI assistant.

Summarize the following text in 2-3 concise sentences:

{text}
"""

# Step 3: Generate summary
summary = ask_llm(prompt)

print("\n=== Summary ===")
print(summary)
```
