```python
from transformers import AutoTokenizer, AutoModelForCausalLM
from geopy.distance import geodesic
import torch
```

## Model Loading

```python
model_name = "Qwen/Qwen2-1.5B-Instruct"

print("Loading model...")

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float32,
    device_map="auto"
)

print("Model loaded!\n")
```

## LLM Function

```python
def ask_llm(prompt):
    
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=150,
        temperature=0.7
    )

    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

## Sample Places

```python
places = {
    "Cairo Tower": (30.0459, 31.2243),
    "Egyptian Museum": (30.0478, 31.2336),
    "Al-Azhar Park": (30.0400, 31.2620)
}
```

## User Location

```python
print("=== AI Distance Agent ===")
user_lat = float(input("Enter your latitude: "))
user_lon = float(input("Enter your longitude: "))

user_location = (user_lat, user_lon)
```

## LLM Model

```python
prompt = f"""
You MUST use the following result:

Closest place: {closest_place}
Distance: {closest_distance:.2f} km

Do NOT change or recompute anything.

Just explain it nicely.
"""

response = ask_llm(prompt)

print("\n=== Recommendation ===")
print(response)
```
