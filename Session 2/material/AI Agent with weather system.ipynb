```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
```

## Loading Model


```python
model_name = "Qwen/Qwen2-1.5B-Instruct"

print("Loading model...")

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float32,
    device_map="auto"
)
print("Model loaded!\n")
```

## Weather Tool

```python
def weather_tool(city):
    
    mock_data = {
        "cairo": {"temp": 36, "condition": "sunny"},
        "london": {"temp": 12, "condition": "rainy"},
        "dubai": {"temp": 40, "condition": "hot"},
    }
    return mock_data.get(city.lower(), {"temp": 25, "condition": "normal"})
```

## LLM Function

```python
def ask_llm(prompt):
    
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=120,
        temperature=0.7
    )

    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

# AI Agent

```python
def weather_agent(city):
    print("[Tool] Getting weather data...")
    data = weather_tool(city)

    temp = data["temp"]
    condition = data["condition"]

    # Control logic (important!)
    if temp > 35:
        decision = "stay indoors"

    elif condition == "rainy":
        decision = "carry an umbrella"
        
    else:
        decision = "go outside"

    print("[Agent] Decision made based on tool")

    prompt = f"""
    You are an AI assistant.

    Weather in {city}:
    Temperature: {temp}°C
    Condition: {condition}
    Recommended action: {decision}

    Explain this recommendation briefly in a friendly way (1-2 sentences only).
    
    """

    return ask_llm(prompt)
```

## Tests

```python
print("=== Weather AI Agent ===")
city = input("Enter a city: ")

result = weather_agent(city)

print("\n=== Recommendation ===")
print(result)
```
